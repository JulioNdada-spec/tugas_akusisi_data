Verifikasi Environment Python:

```python
import sys
import platform

print("Python :", sys.version)
print("Platform:", platform.platform())

for lib in ["pandas", "numpy", "requests", "bs4"]:
    try:
        m = __import__(lib)
        v = getattr(m, "__version__", "terpasang")
        print(f"{lib:10s}: {v}")
    except ImportError:
        print(f"{lib:10s}: BELUM TERPASANG")
```
